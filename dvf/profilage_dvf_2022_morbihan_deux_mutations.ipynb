# DVF 2022 — Deux mutations géantes du Morbihan (27 décembre 2022)

Le département du Morbihan (56) doit son ratio moyenne/médiane très élevé à deux mutations enregistrées le 27 décembre 2022, dont la valeur foncière est répétée sur des milliers de lignes réparties dans de nombreuses communes. Ce notebook décompose chacune de ces deux mutations par type de bien, afin d'observer la diversité des biens auxquels un même prix est associé.

## Cellule 1 — Connexion au fichier

Interrogation avec DuckDB, directement au format Parquet. Le chemin pointe vers le fichier `dvf-2022.parquet` placé dans le dossier `data/`.

```python
import duckdb
from pathlib import Path

# Chemin vers le fichier Parquet, place dans le dossier data/
FICHIER = Path(r"./data/dvf-2022.parquet")

con = duckdb.connect()
pq = str(FICHIER)
assert FICHIER.exists(), f"Fichier introuvable : {FICHIER}"
print(f"Fichier : {FICHIER.name}")

# Les deux montants concernes
VALEUR_A = 337200416
VALEUR_B = 314985152
```

## Cellule 2 — Première mutation : 337 200 416 euros

Répartition des lignes portant ce montant, par type de bien, avec le nombre de communes concernées pour chaque type. Les lignes sans type de local correspondent à des terrains.

```python
mutation_a = con.execute(f"""
    SELECT COALESCE("Type local", '(vide) = terrain') AS type_local,
           COUNT(*)                       AS nb_lignes,
           COUNT(DISTINCT "Commune")      AS nb_communes
    FROM '{pq}'
    WHERE "Code departement" = '56' AND "Valeur fonciere" = {VALEUR_A}
    GROUP BY 1
    ORDER BY nb_lignes DESC
""").fetchdf()

total_a = con.execute(f"""
    SELECT COUNT(*), COUNT(DISTINCT "Commune")
    FROM '{pq}'
    WHERE "Code departement" = '56' AND "Valeur fonciere" = {VALEUR_A}
""").fetchone()

print(f"Mutation a {VALEUR_A:,} EUR : {total_a[0]:,} lignes sur {total_a[1]} communes".replace(',', ' '))
mutation_a
```

## Cellule 3 — Seconde mutation : 314 985 152 euros

Même décomposition pour le second montant. C'est cette mutation qui inclut le groupe de 976 dépendances de Vannes.

```python
mutation_b = con.execute(f"""
    SELECT COALESCE("Type local", '(vide) = terrain') AS type_local,
           COUNT(*)                       AS nb_lignes,
           COUNT(DISTINCT "Commune")      AS nb_communes
    FROM '{pq}'
    WHERE "Code departement" = '56' AND "Valeur fonciere" = {VALEUR_B}
    GROUP BY 1
    ORDER BY nb_lignes DESC
""").fetchdf()

total_b = con.execute(f"""
    SELECT COUNT(*), COUNT(DISTINCT "Commune")
    FROM '{pq}'
    WHERE "Code departement" = '56' AND "Valeur fonciere" = {VALEUR_B}
""").fetchone()

print(f"Mutation a {VALEUR_B:,} EUR : {total_b[0]:,} lignes sur {total_b[1]} communes".replace(',', ' '))
mutation_b
```
