# DVF 2022 — Aperçu des premières lignes du fichier

Ce notebook affiche les premières lignes du fichier DVF 2022, limitées à quelques colonnes lisibles (date, prix, commune, type de local, surfaces). Il permet d'observer la structure multi-lignes : plusieurs lignes peuvent appartenir à une même mutation et porter le même prix.

## Cellule 1 — Connexion au fichier

Interrogation avec DuckDB, directement au format Parquet. Le chemin pointe vers le fichier `dvf-2022.parquet` placé dans le dossier `data/`.

```python
import duckdb
from pathlib import Path

# Chemin vers le fichier Parquet, place dans le dossier data/
FICHIER = Path(r"./data/dvf-2022.parquet")

con = duckdb.connect()
pq = str(FICHIER)
assert FICHIER.exists(), f"Fichier introuvable : {FICHIER}"
nb_lignes = con.execute(f"SELECT count(*) FROM '{pq}'").fetchone()[0]
print(f"Fichier : {FICHIER.name}")
print(f"Lignes  : {nb_lignes:,}".replace(',', ' '))
```

## Cellule 2 — Les dix premières lignes

Sélection des dix premières lignes du fichier, sur six colonnes. Les lignes qui partagent la même date, la même commune et le même prix appartiennent à une même mutation. Les lignes sans type de local sont des parcelles de terrain.

```python
apercu = con.execute(f"""
    SELECT "Date mutation",
           "Valeur fonciere",
           "Commune",
           "Type local",
           "Surface reelle bati",
           "Surface terrain"
    FROM '{pq}'
    LIMIT 10
""").fetchdf()

apercu
```
