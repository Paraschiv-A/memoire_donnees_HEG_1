## complement_renvois_bofip

**Fichier(s) source :** `./data/inventaire_bofip_stock_live_20260521.csv` (inventaire produit par `profilage_bofip_consolide.ipynb`)

**Fichier(s) de sortie :** DataFrame en mémoire

**Description :** Complément à `profilage_bofip_consolide` : mesure du nombre de documents Contenu sans aucun renvoi (dc:relation), avec rappel du total des renvois et des orphelins par type.

```python
import pandas as pd
import os

# Charger l'inventaire produit par profilage_bofip_consolide
CSV = r"./data/inventaire_bofip_stock_live_20260521.csv"

df = pd.read_csv(CSV)
print(f"Documents chargés : {len(df)}")

# Documents sans aucun renvoi
sans_renvoi = (df["nb_renvois"] == 0).sum()
print(f"\nDocuments sans aucun renvoi : {sans_renvoi} ({sans_renvoi/len(df):.1%})")

# Documents avec au moins un orphelin
avec_orphelin = (df["nb_orphelins"] > 0).sum()
print(f"Documents avec au moins un orphelin : {avec_orphelin}")

# Total renvois et orphelins
tot_r = df["nb_renvois"].sum()
tot_o = df["nb_orphelins"].sum()
print(f"Total renvois : {tot_r}")
print(f"Total orphelins : {tot_o} ({tot_o/tot_r:.1%})")
print(f"\nOrphelins par type :")
print(f"  Actualité : {int(df['orph_actualite'].sum())}")
print(f"  Contenu : {int(df['orph_contenu'].sum())}")
print(f"  Fichier : {int(df['orph_fichier'].sum())}")
```
