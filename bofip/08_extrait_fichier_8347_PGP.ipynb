## extrait_fichier_8347_PGP

**Fichier(s) source :** `./data/bofip_stock_live_20260521.tgz` (stock du 21.05.2026)

**Fichier(s) de sortie :** `./data/extrait_8347-PGP/8347-PGP_document.xml` et `./data/extrait_8347-PGP/8347-PGP.html`

**Description :** Extraction des fichiers d'un document BOFiP précis (identifiant 8347-PGP) depuis l'archive : ses métadonnées (document.xml) et son contenu (data.html) sont enregistrés dans un sous-dossier dédié.

```python
import os, tarfile, glob

# --- chemins (repris de l'Etape 1) ---
TGZ_PATH = r"./data/bofip_stock_live_20260521.tgz"
IDENTIFIANT = "8347-PGP"

# --- retrouver les fichiers du document dans l'archive ---
with tarfile.open(TGZ_PATH, "r:gz") as tar:
    membres = [m.name for m in tar.getmembers() if m.isfile()]

def appartient(nom):
    import re
    for s in re.split(r"[\\/]", nom):
        if s == IDENTIFIANT or s.split(".")[0] == IDENTIFIANT:
            return True
    return False

fichiers_doc = [n for n in membres if appartient(n)]
print("Fichiers trouves :", len(fichiers_doc))

# --- enregistrer contenu + metadonnees ---
dossier = os.path.join(os.path.dirname(TGZ_PATH), "extrait_" + IDENTIFIANT)
os.makedirs(dossier, exist_ok=True)

with tarfile.open(TGZ_PATH, "r:gz") as tar:
    for n in fichiers_doc:
        ext = os.path.splitext(n)[1].lower()
        data = tar.extractfile(n).read()
        if ext == ".xml":
            cible = os.path.join(dossier, IDENTIFIANT + "_document.xml")
        else:
            cible = os.path.join(dossier, IDENTIFIANT + (ext if ext else ".html"))
        with open(cible, "wb") as f:
            f.write(data)
        print("Enregistre :", cible)

print()
print("Dossier complet :", os.path.abspath(dossier))
try:
    os.startfile(dossier)
except Exception:
    pass
```
