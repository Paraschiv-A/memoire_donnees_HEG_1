## inspecter_6_sans_sujet

**Fichier(s) source :** `./data/bofip_stock_live_20260521.tgz` (stock du 21.05.2026)

**Fichier(s) de sortie :** Aucun (résultats affichés en sortie)

**Description :** Inspection ciblée des 6 documents identifiés comme dépourvus de métadonnée dc:subject (identifiants 7362-PGP à 7367-PGP) : extraction et affichage de leur contenu depuis l'archive BOFiP.

```python
import tarfile, os

ARCHIVE = r"./data/bofip_stock_live_20260521.tgz"
IDS = {"7362-PGP", "7363-PGP", "7364-PGP", "7365-PGP", "7366-PGP", "7367-PGP"}

trouve = {}

with tarfile.open(ARCHIVE, "r:gz") as tar:
    for m in tar.getmembers():
        if not m.isfile():
            continue
        chemin = m.name.replace("\\", "/")
        for code in IDS:
            if "/" + code + "/" in chemin:
                f = tar.extractfile(m)
                if f is None:
                    continue
                contenu = f.read().decode("utf-8", "replace")
                nom = os.path.basename(chemin)
                cle = f"{code}/{nom}"
                if cle not in trouve:
                    trouve[cle] = contenu
                    print("=" * 70)
                    print(f"{code} / {nom} ({len(contenu)} caractères)")
                    print("=" * 70)
                    print(contenu[:800])
                    if len(contenu) > 800:
                        print(f"... (tronqué)")
                    print()
                break

if not trouve:
    print("Aucun document trouvé. Vérifier les identifiants.")
else:
    print(f"\n{len(trouve)} fichiers trouvés pour {len(IDS)} identifiants.")
```
