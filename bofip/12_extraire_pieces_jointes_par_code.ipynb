## extraire_pieces_jointes_par_code

**Fichier(s) source :** `./data/bofip_stock_live_20260521.tgz` (stock du 21.05.2026)

**Fichier(s) de sortie :** un sous-dossier `./data/extrait_<CODE>/` par code (pièces jointes extraites), pour 5466-PGP, 10838-PGP, 11094-PGP et 13100-PGP

**Description :** Extraction des pièces jointes (fichiers liés) de plusieurs documents BOFiP désignés par leur code, chacune enregistrée dans un sous-dossier dédié.

```python
import os, tarfile

TGZ_PATH = r"./data/bofip_stock_live_20260521.tgz"
CODES = ["5466-PGP", "10838-PGP", "11094-PGP", "13100-PGP"]

with tarfile.open(TGZ_PATH, "r:gz") as tar:
    membres = [m.name for m in tar.getmembers() if m.isfile()]

for CODE in CODES:
    cibles = [n for n in membres if "/" + CODE + "/" in n.replace("\\", "/")]
    if not cibles:
        print(CODE, ": aucun fichier"); continue
    dossier = os.path.join(os.path.dirname(TGZ_PATH), "extrait_" + CODE)
    os.makedirs(dossier, exist_ok=True)
    with tarfile.open(TGZ_PATH, "r:gz") as tar:
        for n in cibles:
            nom = os.path.basename(n) or (CODE + ".bin")
            ext = ".pdf" if (nom == "data1" or "." not in nom) else ""
            with open(os.path.join(dossier, nom + ext), "wb") as f:
                f.write(tar.extractfile(n).read())
    print(CODE, ": extrait dans", os.path.basename(dossier))
```
